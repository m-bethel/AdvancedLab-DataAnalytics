```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def load_and_clean(filename, skip=1):
    df = pd.read_csv(filename, encoding='latin-1', skiprows=skip)
    df = df.dropna(axis=1, how='all')
    df = df.dropna(subset=[df.columns[1]]) 
    new_names = ['time', 'CPM', 'error', 'Offset']
    df.columns = new_names[:len(df.columns)]
    df["counts"] = df["CPM"].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
    
    return df

# --- Load the datasets ---
bg = load_and_clean("PH336_Background_Count_Data_2.7.csv", skip=5)
d1 = load_and_clean("PH336_Experiment_Count_Data_2.7.csv", skip=1)
d2 = load_and_clean("PH336_Experiment_Count_Data_3.0.csv", skip=1)
d3 = load_and_clean("PH336_Experiment_Count_Data_4.0.csv", skip=1)
d4 = load_and_clean("PH336_Experiment_Count_Data_1.0.csv", skip=1)
d5 = load_and_clean("PH336_Experiment_Count_Data_5.0.csv", skip=1)
d6 = load_and_clean("PH336_Experiment_Count_Data_6.0.csv", skip=1)

# Finding background with the minimum value across each set
bg_min = np.array([
    d4["counts"].min(),
    d1["counts"].min(),
    d2["counts"].min(),
    d3["counts"].min(),
    d5["counts"].min(),
    d6["counts"].min()]).min()

bg_max = np.array([
    d4["counts"].max(),
    d1["counts"].max(),
    d2["counts"].max(),
    d3["counts"].max(),
    d5["counts"].max(),
    d6["counts"].max()]).max()

# Subtract background to get the TRUE positron signal
avg_4 = d4["counts"].mean()-bg_min
avg_1 = d1["counts"].mean()-bg_min
avg_2 = d2["counts"].mean()-bg_min
avg_3 = d3["counts"].mean()-bg_min
avg_5 = d5["counts"].mean()-bg_min
avg_6 = d6["counts"].mean()-bg_min

#print(f"Net Average 2.7cm: {avg_1:.2f}")
#print(f"Net Average 3.0cm: {avg_2:.2f}")
#print(f"Net Average 4.0cm: {avg_3:.2f}")
#print(f"Net Average 1.0cm: {avg_4:.2f}")

x_data = np.array([1.0, 2.7, 4.0, 5.0, 6.0]) #discluded 3.0
y_data = np.array([avg_4, avg_1, avg_2, avg_5, avg_6]) # discluded: , avg_3
y_err = np.array([
    d4["counts"].sem(),
    d1["counts"].sem(),
    d2["counts"].sem(),
    #d3["counts"].sem(),
    d5["counts"].sem(),
    d6["counts"].sem()])

def gaussian(x, a, mu, sigma):
    return a * np.exp(-(x - mu)**2 / (2 * sigma**2))

guesses=[bg_max-bg_min, 3.5, 0.1]

upper_b = [5,5.0,1.5]
lower_b = [0,0,0]

params, pcov = curve_fit(gaussian, x_data, y_data, sigma=y_err,
                         p0=guesses, bounds=(lower_b, upper_b))

a_fit, mu_fit, sigma_fit = params
mu_error = np.sqrt(pcov[1,1])
print(f"The beam center is {mu_fit:.3f} cm.")
# Plotting
x_fine = np.linspace(0, 7, 500)
plt.errorbar(x_data, y_data, yerr=y_err, fmt='o', color='black', label='Data with Error Bars')
plt.plot(x_fine, gaussian(x_fine, *params), 'r-', label=f'Fit Peak: {mu_fit:.2f} ± {mu_error:.2f} cm')

plt.title("Gaussian Fit of Positron Beam")
plt.xlabel("Deflection (cm)")
plt.ylabel("Counts")
plt.legend()
plt.show()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import itertools # Added for the grid search loop

def load_and_clean(filename, skip=1):
    df = pd.read_csv(filename, encoding='latin-1', skiprows=skip)
    df = df.dropna(axis=1, how='all')
    df = df.dropna(subset=[df.columns[1]]) 
    new_names = ['time', 'CPM', 'error', 'Offset']
    df.columns = new_names[:len(df.columns)]
    df["counts"] = df["CPM"].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
    return df

# --- Load the datasets ---
bg = load_and_clean("PH336_Background_Count_Data_2.7.csv", skip=5)
d1 = load_and_clean("PH336_Experiment_Count_Data_2.7.csv", skip=1)
d2 = load_and_clean("PH336_Experiment_Count_Data_3.0.csv", skip=1)
d3 = load_and_clean("PH336_Experiment_Count_Data_4.0.csv", skip=1)
d4 = load_and_clean("PH336_Experiment_Count_Data_1.0.csv", skip=1)
d5 = load_and_clean("PH336_Experiment_Count_Data_5.0.csv", skip=1)
d6 = load_and_clean("PH336_Experiment_Count_Data_6.0.csv", skip=1)

# Finding background by finding the minimum value across each set
bg_min = np.array([
    d4["counts"].min(),
    d1["counts"].min(),
    d2["counts"].min(),
    d3["counts"].min(),
    d5["counts"].min(),
    d6["counts"].min()]).min() # Fixed the duplicate d3 from original code

# Subtract background to get the TRUE positron signal
avg_4 = d4["counts"].mean() - bg_min
avg_1 = d1["counts"].mean() - bg_min
avg_2 = d2["counts"].mean() - bg_min
avg_3 = d3["counts"].mean() - bg_min
avg_5 = d5["counts"].mean() - bg_min
avg_6 = d6["counts"].mean() - bg_min

x_data = np.array([1.0, 2.7, 3.0, 4.0, 5.0, 6.0])
y_data = np.array([avg_4, avg_1, avg_2, avg_3, avg_5, avg_6])
y_err = np.array([
    d4["counts"].sem(),
    d1["counts"].sem(),
    d2["counts"].sem(),
    d3["counts"].sem(),
    d5["counts"].sem(),
    d5["counts"].sem()])

def gaussian(x, a, mu, sigma):
    return a * np.exp(-(x - mu)**2 / (2 * sigma**2))

# =====================================================================
# --- GRID SEARCH FOR BEST INITIAL GUESS (Least Squares Loop) ---
# =====================================================================

# Define the search space (min, max, number of steps)
a_grid = np.linspace(2.4, 2.6, 25)  # Amplitude guess
mu_grid = np.linspace(3.6, 3.9, 25)               # Center guess
sigma_grid = np.linspace(1.35, 1.55, 25)            # Width guess

best_ssr = np.inf
best_guesses = [1.0, 3.0, 0.5]

# Iterate through every possible combination of a, mu, and sigma
for a_g, mu_g, sigma_g in itertools.product(a_grid, mu_grid, sigma_grid):
    # Calculate what the curve WOULD look like
    y_pred = gaussian(x_data, a_g, mu_g, sigma_g)
    
    # Calculate the Weighted Sum of Squared Residuals (Chi-Squared)
    # We divide by the error so points with high uncertainty matter less
    weights = 1.0 / (y_err**2 + 1e-10) # 1e-10 prevents division by zero
    ssr = np.sum(weights * (y_data - y_pred)**2)
    
    # If this is the lowest error we've seen, save it!
    if ssr < best_ssr:
        best_ssr = ssr
        best_guesses = [a_g, mu_g, sigma_g]

print(f"Grid Search Best Guesses: Amplitude={best_guesses[0]:.1f}, Center={best_guesses[1]:.2f}, Sigma={best_guesses[2]:.2f}")

# =====================================================================
# --- FINAL CURVE FIT ---
# =====================================================================

# Opened up the upper bounds so the fit doesn't flatline!
lower_b = [0, 0, 0]
upper_b = [max(y_data) * 3, 7.0, 3.0] 

# Pass the best_guesses from the loop directly into p0
params, pcov = curve_fit(
    gaussian, 
    x_data, 
    y_data, 
    sigma=y_err,
    p0=best_guesses, 
    bounds=(lower_b, upper_b),
    absolute_sigma=True)

a_fit, mu_fit, sigma_fit = params
mu_error = np.sqrt(pcov[1,1])

print(f"Final Optimized Beam Center is {mu_fit:.3f} ± {mu_error:.3f} cm.")

# --- Plotting ---
x_fine = np.linspace(0, 7, 500)
plt.figure(figsize=(8, 5))
plt.errorbar(x_data, y_data, yerr=y_err, fmt='o', color='black', label='Error Bar', zorder=5)
plt.plot(x_fine, gaussian(x_fine, *params), 'r-', label=f'Fit Peak: {mu_fit:.2f} ± {mu_error:.2f} cm')

plt.title("Gaussian Fit of Positron Beam")
plt.xlabel("Deflection (cm)")
plt.ylabel("Counts")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np

# --- 1. Constants ---
q = 1.602e-19
c = 299792458
m_accepted = 9.109e-31

# --- 2. Input Your Measurements and Uncertainties ---
# Order: [Deflection (y), Voltage (V), B-field (B), Gap (d), Field Length (Lf), Total Length (Lt)]

# The actual measurements
vals = [mu_fit, 980, 0.00435, 0.001, 0.12, 0.18]

# Uncertainties for each measurement
errs = [
    0.001,    #y (Gaussian mu_error in meters)
    5,       #V (Voltage wobble)
    0.00005,  #B (Gauss meter precision in Tesla)
    0.0001,   #d (Capacitor gap error in meters)
    0.005,    #Lf (Effective coil field length error in meters)
    0.002]    #Lt (Detector distance error in meters)


# --- 3. The Core Physics Function ---
def calc_mass(inputs):
    y, V, B, d, Lf, Lt = inputs    
    R = (Lf * (2 * Lt - Lf)) / (2 * y)
    v = V / (B * d)   
    # Safety check
    if v >= c: return np.nan 
    gamma = 1 / np.sqrt(1 - (v/c)**2)
    return (q * B * R) / (gamma * v)

# --- 4. Numerical Error Propagation ---
m_nominal = calc_mass(vals)
variance_sum = 0

# Loop through each variable to calculate its partial derivative
for i in range(len(vals)):
    # Create a tiny nudge (epsilon) to find the slope
    epsilon = vals[i] * 1e-6 if vals[i] != 0 else 1e-8
    
    # Nudge the variable UP
    vals_up = list(vals)
    vals_up[i] += epsilon
    m_up = calc_mass(vals_up)
    
    # Nudge the variable DOWN
    vals_down = list(vals)
    vals_down[i] -= epsilon
    m_down = calc_mass(vals_down)
    
    # Calculate the partial derivative (slope) for this specific variable
    partial_derivative = (m_up - m_down) / (2 * epsilon)
    
    # Add its contribution to the total variance: (df/dx * dx)^2
    variance_sum += (partial_derivative * errs[i])**2

# Final standard error is the square root of the variance
m_error = np.sqrt(variance_sum)

# --- 5. Final Output ---
diff = abs(m_nominal - m_accepted)
z_score = diff / m_error

print("\n--- Error Propagation Results ---")
print(f"Experimental Mass:      {m_nominal:.2e} kg")
print(f"Absolute Uncertainty: ± {m_error:.2e} kg")
print("")
print(f"Difference from true:   {diff:.2e} kg")
print(f"Sigma value:            {z_score:.2f} sigma")

if z_score <= 2.0:
    print("\nConclusion: SUCCESS! The accepted mass is within your margin of error.")
else:
    print("\nConclusion: The accepted mass is outside your error margin.")
```
